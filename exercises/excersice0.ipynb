```python
df_quantity_year = df_quantity.groupby("Year").sum()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df_quantity_year.index, df_quantity_year["Quantity"], marker='o')
plt.title("Total Car Sales in Norway (by Year)")
plt.xlabel("Year")
plt.ylabel("Number of Cars Sold")
plt.grid()
plt.show()
```
