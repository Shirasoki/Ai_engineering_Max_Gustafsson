# Testing a post request in Python

```python
import requests

url = "http://127.0.0.1:8000/api/predict"
payload ={
  "SepalLengthCm": 6.1,
  "SepalWidthCm": 2.8,
  "PetalLengthCm": 4.3,
  "PetalWidthCm": 1.3
}
response = requests.post(url, json=payload)
response
```

```python
response.json()
```
